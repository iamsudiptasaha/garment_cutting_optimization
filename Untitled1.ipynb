```python
def render_marker(
    placed_shapes,
    strip_width,
    title=None,
    save_path=None,      # NEW
    dpi=300              # NEW
):
    """
    Render a fabric marker layout with part + size labels.
¯
    save_path: str or None
        e.g. "marker.png" or "marker.pdf"
    """

    fig, ax = plt.subplots(figsize=(10, 14))

    patches = []
    colors = []
    part_colors = {}

    def get_color(part):
        if part not in part_colors:
            part_colors[part] = (
                random.random(),
                random.random(),
                random.random(),
                0.6
            )
        return part_colors[part]

    max_y = 0.0

    for item in placed_shapes:
        poly = item["polygon"]
        part = item["part"]
        size = item["size"]

        patch = Polygon(poly, closed=True)
        patches.append(patch)
        colors.append(get_color(part))

        # centroid label
        cx = sum(p[0] for p in poly) / len(poly)
        cy = sum(p[1] for p in poly) / len(poly)

        ax.text(
            cx, cy,
            f"{part}\n{size}",
            ha="center",
            va="center",
            fontsize=8,
            fontweight="bold"
        )

        for _, y in poly:
            max_y = max(max_y, y)

    collection = PatchCollection(
        patches,
        facecolors=colors,
        edgecolors="black",
        linewidths=0.8
    )
    ax.add_collection(collection)

    # Fabric boundary
    ax.plot([0, strip_width], [0, 0], color="black", linewidth=2)
    ax.plot([0, strip_width], [max_y, max_y], color="black", linewidth=2)
    ax.plot([0, 0], [0, max_y], color="black", linewidth=2)
    ax.plot([strip_width, strip_width], [0, max_y], color="black", linewidth=2)

    ax.set_xlim(0, strip_width)
    ax.set_ylim(max_y, 0)
    ax.set_aspect("equal")
    ax.set_xlabel("Fabric Width (in)")
    ax.set_ylabel("Fabric Length (in)")

    if title:
        ax.set_title(title)

    plt.tight_layout()

    # ✅ SAVE HERE
    if save_path:
        plt.savefig(save_path, dpi=dpi, bbox_inches="tight")

    plt.show()
    plt.close(fig)
```

```python
MM_TO_INCH = 1.0 / 25.4

def path_to_polygon(path, samples_per_curve=40):
    raw = []

    for segment in path:
        for t in np.linspace(0, 1, samples_per_curve):
            pt = segment.point(t)
            raw.append((pt.real, pt.imag))

    # 1. Convert mm → inches
    pts = [(x * MM_TO_INCH, y * MM_TO_INCH) for x, y in raw]

    # 2. Flip Y axis (SVG down → math up)
    ys = [y for _, y in pts]
    max_y = max(ys)
    pts = [(x, max_y - y) for x, y in pts]

    # 3. Normalize origin to (0,0)
    xs = [x for x, _ in pts]
    ys = [y for _, y in pts]

    min_x = min(xs)
    min_y = min(ys)

    pts = [(x - min_x, y - min_y) for x, y in pts]

    return pts
```

```python
def load_svg_layer_as_polygon(svg_file, layer_name, samples_per_curve=40):
    paths, attributes, svg_attr = svg2paths2(svg_file)

    # --- Parse viewBox ---
    vb = list(map(float, svg_attr["viewBox"].split()))
    vb_x, vb_y, vb_w, vb_h = vb

    # --- Parse physical size (mm) ---
    width_mm = float(svg_attr["width"].replace("mm", ""))
    height_mm = float(svg_attr["height"].replace("mm", ""))

    scale_x = width_mm / vb_w
    scale_y = height_mm / vb_h

    for path, attr in zip(paths, attributes):
        label = (
            attr.get("inkscape:label")
            or attr.get("label")
            or attr.get("id")
        )

        if label and layer_name in label:
            raw = []
            for segment in path:
                for t in np.linspace(0, 1, samples_per_curve):
                    pt = segment.point(t)
                    raw.append((pt.real, pt.imag))

            # Apply viewBox → mm → inches
            pts = []
            for x, y in raw:
                x_mm = (x - vb_x) * scale_x
                y_mm = (y - vb_y) * scale_y
                pts.append((x_mm / 25.4, y_mm / 25.4))

            # Flip Y and normalize
            max_y = max(y for _, y in pts)
            pts = [(x, max_y - y) for x, y in pts]

            min_x = min(x for x, _ in pts)
            min_y = min(y for _, y in pts)
            pts = [(x - min_x, y - min_y) for x, y in pts]

            return pts

    raise ValueError(f"Layer '{layer_name}' not found")
```
