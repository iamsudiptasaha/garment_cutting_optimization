```python
import math
import copy
import os
import re
import json
import random
from collections import defaultdict
import xml.etree.ElementTree as ET
from collections import defaultdict

from shapely.geometry import Polygon as ShapelyPolygon, MultiPolygon, box
from shapely.affinity import translate, rotate
from shapely.ops import unary_union
from shapely.geometry import box
from shapely.affinity import translate

import matplotlib.pyplot as plt
from matplotlib.patches import Polygon as MplPolygon
from matplotlib.collections import PatchCollection

from svgpathtools import svg2paths2

import numpy as np



```

```python
def draw_geometry(ax, geom, color, alpha=0.3):
    patches = []

    if geom.is_empty:
        return

    if geom.geom_type == "Polygon":
        patches.append(MplPolygon(list(geom.exterior.coords)))
    elif geom.geom_type == "MultiPolygon":
        for g in geom.geoms:
            patches.append(MplPolygon(list(g.exterior.coords)))

    pc = PatchCollection(
        patches,
        facecolor=color,
        edgecolor="none",
        alpha=alpha
    )
    ax.add_collection(pc)

def debug_free_space_snapshot(
    placed_polys,
    strip_width,
    iter_idx,
    save_dir="debug"
):
    import os
    os.makedirs(save_dir, exist_ok=True)

    fig, ax = plt.subplots(figsize=(6, 12))

    # ---- Occupied ----
    occupied = None
    if placed_polys:
        occupied = safe_union(placed_polys)
        draw_geometry(ax, occupied, color="black", alpha=0.8)

    # ---- Free space ----
    free_space = compute_free_space(placed_polys, strip_width)
    draw_geometry(ax, free_space, color="cyan", alpha=0.3)

    # ---- Strip boundary ----
    max_y = max(p.bounds[3] for p in placed_polys) + 5 if placed_polys else 50
    ax.plot(
        [0, strip_width, strip_width, 0, 0],
        [0, 0, max_y, max_y, 0],
        color="black",
        linewidth=1
    )

    ax.set_xlim(0, strip_width)
    ax.set_ylim(0, max_y)
    ax.set_title(f"Free Space Debug – Iter {iter_idx}")
    ax.set_aspect("equal")
    ax.axis("off")

    plt.savefig(f"{save_dir}/iter_{iter_idx:02d}.png", dpi=150)
    plt.close()


def render_free_space_debug(
    placed_polys,
    strip_width,
    iteration,
    folder,
    final=False
):
    os.makedirs(folder, exist_ok=True)

    free = compute_free_space(placed_polys, strip_width)
    cavities = extract_cavities(free)

    fig, ax = plt.subplots(figsize=(6,10))

    # ---- draw free space ----
    if not free.is_empty:
        if free.geom_type == "Polygon":
            xs, ys = free.exterior.xy
            ax.fill(xs, ys, color="#8fd3d3", alpha=0.6)
        else:
            for g in free.geoms:
                xs, ys = g.exterior.xy
                ax.fill(xs, ys, color="#8fd3d3", alpha=0.6)

    # ---- draw cavities ----
    colors = ["#ff9999","#ffd27f","#a0ffb3","#a0c4ff","#d0a0ff"]
    for i, cav in enumerate(cavities):
        xs, ys = cav.exterior.xy
        ax.fill(xs, ys, color=colors[i%len(colors)], alpha=0.5)

    # ---- draw placed pieces ----
    for p in placed_polys:
        xs, ys = p.exterior.xy
        ax.fill(xs, ys, color="#2b2b2b")

    ax.set_xlim(0, strip_width)
    ax.set_ylim(0, max(50, max([p.bounds[3] for p in placed_polys], default=10)+10))
    ax.set_title(f"Cavity Debug Iter {iteration}")
    ax.set_aspect("equal")

    fname = f"debug_iter_{iteration}.png"
    if final:
        fname = "debug_final_cavities.png"

    plt.savefig(os.path.join(folder, fname), dpi=150, bbox_inches="tight")
    plt.close()
```

```python
def extract_color(attr):
    """
    Extract fill or stroke color from SVG attributes.
    Supports hex and rgb() styles from CorelDRAW.
    """

    # Direct attributes
    if attr.get("fill") and attr["fill"] != "none":
        return attr["fill"]

    if attr.get("stroke") and attr["stroke"] != "none":
        return attr["stroke"]

    style = attr.get("style", "")

    # Hex colors
    m = re.search(r"fill:\s*(#[0-9A-Fa-f]{6})", style)
    if m:
        return m.group(1)

    m = re.search(r"stroke:\s*(#[0-9A-Fa-f]{6})", style)
    if m:
        return m.group(1)

    # rgb() colors (Corel)
    m = re.search(r"(fill|stroke):\s*rgb\([^)]+\)", style)
    if m:
        return m.group(0)  # use full rgb(...) string as key

    return None

def parse_filename(filename):
    """
    JB_FRONT_16-18-20.svg  → sizes = ['16','18','20']
    JB_FRONT_M-L-XL.svg    → sizes = ['M','L','XL']
    """

    name = os.path.splitext(filename)[0]

    parts = name.split("_")
    if len(parts) < 3:
        raise ValueError(f"Invalid filename format: {filename}")

    brand = parts[0]
    part = parts[1].upper()
    size_list = parts[2]

    sizes = size_list.split("-")

    return part, sizes

def svg_path_to_polygon(path, svg_attr, samples=40):
    # --- Parse viewBox ---
    vb_x, vb_y, vb_w, vb_h = map(float, svg_attr["viewBox"].split())

    # --- Parse physical size in mm ---
    width_mm = float(svg_attr["width"].replace("mm", ""))
    height_mm = float(svg_attr["height"].replace("mm", ""))

    scale_x = width_mm / vb_w
    scale_y = height_mm / vb_h

    raw = []
    for seg in path:
        for t in np.linspace(0, 1, samples):
            pt = seg.point(t)
            raw.append((pt.real, pt.imag))

    # Convert to inches
    pts = []
    for x, y in raw:
        x_mm = (x - vb_x) * scale_x
        y_mm = (y - vb_y) * scale_y
        pts.append((x_mm / 25.4, y_mm / 25.4))

    # Flip Y and normalize
    max_y = max(y for _, y in pts)
    pts = [(x, max_y - y) for x, y in pts]

    min_x = min(x for x, _ in pts)
    min_y = min(y for _, y in pts)
    pts = [(x - min_x, y - min_y) for x, y in pts]

    return pts

def load_svg_file(svg_path):
    paths, attrs, svg_attr = svg2paths2(svg_path)

    colored_groups = defaultdict(list)
    uncolored_paths = []

    for path, attr in zip(paths, attrs):
        color = extract_color(attr)
        if color is None:
            uncolored_paths.append(path)
        else:
            colored_groups[color].append(path)

    polygons = []

    # --- Case 1: color-based grouping ---
    if colored_groups:
        for color, group in colored_groups.items():
            pts = []
            for p in group:
                pts.extend(svg_path_to_polygon(p, svg_attr))

            poly = ShapelyPolygon(pts)

            if poly.area > 1e-3:
                polygons.append(poly)

    # --- Case 2: fallback (no colors) ---
    else:
        for p in paths:
            pts = svg_path_to_polygon(p, svg_attr)
            poly = ShapelyPolygon(pts)

            if poly.area > 1e-3:
                polygons.append(poly)
    
    if not polygons:
        raise ValueError(f"No valid polygons found in {svg_path}")

    return polygons

def assign_sizes(polygons, sizes):
    assert len(polygons) == len(sizes), (
        f"Size count mismatch: {len(polygons)} polygons vs {len(sizes)} sizes"
    )

    areas = [(poly.area, poly) for poly in polygons]
    areas.sort(key=lambda x: x[0])  # smallest → largest
    validate_monotonic_areas(polygons)
    result = []
    for size, (_, poly) in zip(sizes, areas):
        result.append((size, list(poly.exterior.coords)))

    return result

def validate_monotonic_areas(polygons):
    areas = sorted(poly.area for poly in polygons)
    for i in range(len(areas) - 1):
        ratio = areas[i + 1] / areas[i]
        if ratio < 1.02:
            raise ValueError(
                "Two sizes have nearly identical area. "
                "Check colors or duplicate patterns."
            )
def process_design_folder(folder_path):
    all_shapes = []

    for file in os.listdir(folder_path):
        if not file.lower().endswith(".svg"):
            continue

        part, sizes = parse_filename(file)
        svg_path = os.path.join(folder_path, file)

        polygons = load_svg_file(svg_path)
        sized_polys = assign_sizes(polygons, sizes)

        for size, poly in sized_polys:
            all_shapes.append({
                "part": part,
                "size": size,
                "polygon": poly
            })

    # Save structured output
    out_path = os.path.join(folder_path, "parsed_shapes.json")
    with open(out_path, "w") as f:
        json.dump(all_shapes, f, indent=2)

    return all_shapes

def evaluate_kpis(placed_shapes, strip_width):
    max_y = 0.0
    used_area = 0.0

    for item in placed_shapes:
        poly = item["polygon"]

        for _, y in poly:
            max_y = max(max_y, y)

        area = 0.0
        n = len(poly)
        for i in range(n):
            x1, y1 = poly[i]
            x2, y2 = poly[(i + 1) % n]
            area += (x1 * y2 - x2 * y1)
        used_area += abs(area) / 2.0

    total_area = strip_width * max_y
    empty_ratio = 1.0 - (used_area / total_area) if total_area > 0 else 0.0

    return {
        "layer_length_in": max_y,
        "used_area_in2": used_area,
        "total_area_in2": total_area,
        "empty_ratio": empty_ratio
    }
def rotate_180(polygon):
    """
    Rotate polygon 180° around origin and normalize to (0,0).
    """
    rotated = [(-x, -y) for x, y in polygon]

    min_x = min(x for x, _ in rotated)
    min_y = min(y for _, y in rotated)

    normalized = [(x - min_x, y - min_y) for x, y in rotated]
    return normalized

def poly_width_height(poly):
    minx, miny, maxx, maxy = poly.bounds
    return maxx - minx, maxy - miny
    
def clean_polygon(poly, eps=1e-6):
    """
    Makes polygon GEOS-safe by:
    - snapping coordinates
    - fixing self-intersections
    """
    try:
        return poly.buffer(0)
    except Exception:
        return poly.buffer(eps)
```

```python
def safe_union(polys):
    if not polys:
        return None
    fixed = [p.buffer(0) for p in polys]
    return unary_union(fixed).buffer(0)


def envelope_area(polys, strip_width):
    if not polys:
        return 0.0
    max_y = max(p.bounds[3] for p in polys)
    return strip_width * max_y

def compute_free_space(polys, strip_width):
    if not polys:
        return box(0, 0, strip_width, 200)  # initial height

    max_y = max(p.bounds[3] for p in polys) + 2
    strip = box(0, 0, strip_width, max_y)

    occ = safe_union(polys)
    free = strip.difference(occ)
    return free.buffer(0)



def free_space_candidates(free_space):
    pts = []

    if free_space.is_empty:
        return pts

    if free_space.geom_type == "Polygon":
        pts.extend(list(free_space.exterior.coords))
    else:  # MultiPolygon
        for g in free_space.geoms:
            pts.extend(list(g.exterior.coords))

    # unique + sorted (bottom-left bias)
    pts = list(set((round(x, 4), round(y, 4)) for x, y in pts))
    pts.sort(key=lambda p: (p[1], p[0]))
    return pts

def fits(poly, placed, strip_width):
    minx, miny, maxx, maxy = poly.bounds
    if minx < 0 or maxx > strip_width:
        return False
    for p in placed:
        if poly.intersects(p):
            return False
    return True

def extract_cavities(free_space, min_area=1.0):
    cavities = []

    if free_space.is_empty:
        return cavities

    if free_space.geom_type == "Polygon":
        geoms = [free_space]
    else:
        geoms = list(free_space.geoms)

    for g in geoms:
        if g.area >= min_area:
            cavities.append(g)

    return cavities
```

```python
def phase11_envelope_delta_packer(shapes, strip_width):
    placed = []

    for item in shapes:
        base = clean_polygon(ShapelyPolygon(item["polygon"]))
        minx, miny, _, _ = base.bounds

        free_space = compute_free_space(placed, strip_width)
        candidates = free_space_candidates(free_space)

        best = None
        best_score = (float("inf"), float("inf"), float("inf"))

        base_env_area = envelope_area(placed, strip_width)

        for x, y in candidates:
            candidate = translate(
                base,
                xoff=x - minx,
                yoff=y - miny
            )

            if not fits(candidate, placed, strip_width):
                continue

            new_env_area = envelope_area(placed + [candidate], strip_width)
            delta = new_env_area - base_env_area

            top = candidate.bounds[3]

            score = (delta, top, x)

            if score < best_score:
                best_score = score
                best = candidate

        # Fallback: forced placement above
        if best is None:
            y_top = max(p.bounds[3] for p in placed) if placed else 0
            best = translate(base, xoff=-minx, yoff=y_top - miny)

        placed.append(best)
        debug_free_space_snapshot(placed, strip_width, iter_idx=len(placed))

    return placed
```

```python
def phase12_cavity_vision(shapes, strip_width, design_folder):

    placed = []
    y_cursor = 0

    for i, item in enumerate(shapes):
        poly = ShapelyPolygon(item["polygon"]).buffer(0)
        minx, miny, maxx, maxy = poly.bounds

        placed_poly = translate(
            poly,
            xoff=0 - minx,
            yoff=y_cursor - miny
        ).buffer(0)

        placed.append(placed_poly)

        # save iteration debug
        render_free_space_debug(
            placed,
            strip_width,
            iteration=i,
            folder=design_folder
        )

        y_cursor += (maxy - miny) + 0.5

    # final cavity map
    render_free_space_debug(
        placed,
        strip_width,
        iteration="final",
        folder=design_folder,
        final=True
    )

    return placed
```

```python
def render_marker(
    placed_shapes,
    strip_width,
    title=None,
    save_path=None,      # NEW
    dpi=300              # NEW
):
    """
    Render a fabric marker layout with part + size labels.

    save_path: str or None
        e.g. "marker.png" or "marker.pdf"
    """

    fig, ax = plt.subplots(figsize=(10, 14))

    patches = []
    colors = []
    part_colors = {}

    def get_color(part):
        if part not in part_colors:
            part_colors[part] = (
                random.random(),
                random.random(),
                random.random(),
                0.6
            )
        return part_colors[part]

    max_y = 0.0

    for item in placed_shapes:
        poly = item["polygon"]
        part = item["part"]
        size = item["size"]

        patch = MplPolygon(poly, closed=True)

        patches.append(patch)
        colors.append(get_color(part))

        # centroid label
        cx = sum(p[0] for p in poly) / len(poly)
        cy = sum(p[1] for p in poly) / len(poly)

        ax.text(
            cx, cy,
            f"{part}\n{size}",
            ha="center",
            va="center",
            fontsize=8,
            fontweight="bold"
        )

        for _, y in poly:
            max_y = max(max_y, y)

    collection = PatchCollection(
        patches,
        facecolors=colors,
        edgecolors="black",
        linewidths=0.8
    )
    ax.add_collection(collection)

    # Fabric boundary
    ax.plot([0, strip_width], [0, 0], color="black", linewidth=2)
    ax.plot([0, strip_width], [max_y, max_y], color="black", linewidth=2)
    ax.plot([0, 0], [0, max_y], color="black", linewidth=2)
    ax.plot([strip_width, strip_width], [0, max_y], color="black", linewidth=2)

    ax.set_xlim(0, strip_width)
    ax.set_ylim(0, max_y)
    ax.set_aspect("equal")
    ax.set_xlabel("Fabric Width (in)")
    ax.set_ylabel("Fabric Length (in)")

    if title:
        ax.set_title(title)

    plt.tight_layout()

    # ✅ SAVE HERE
    if save_path:
        plt.savefig(save_path, dpi=dpi, bbox_inches="tight")

    plt.show()
    plt.close(fig)
```

```python
STRIP_WIDTH = 36.0
design_folder = "JK-123"

# Load shapes (SVG-driven input)
shapes = process_design_folder(design_folder)

# Run V11
placed_v11_polys = phase11_envelope_delta_packer(
    shapes,
    STRIP_WIDTH
)

# Convert to render/KPI format
placed_v11_items = []
for item, poly in zip(shapes, placed_v11_polys):
    placed_v11_items.append({
        "polygon": list(poly.exterior.coords),
        "part": item["part"],
        "size": item["size"]
    })

# KPIs
kpi_v11 = evaluate_kpis(placed_v11_items, STRIP_WIDTH)

print("V11 KPIs (Envelope-Delta Free-Space):")
for k, v in kpi_v11.items():
    print(f"  {k}: {v:.3f}")
```

```python
render_marker(
    placed_v11_items,
    strip_width=STRIP_WIDTH,
    title="V11 – Envelope-Delta Free-Space Placement",
    save_path=f"{design_folder}/v11_envelope_delta.png"
)
```
